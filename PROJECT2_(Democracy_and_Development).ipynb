```python
import requests
import pandas as pd

countries_wb = {
    "United States": "USA",
    "China": "CHN",
    "Germany": "DEU",
    "Brazil": "BRA",
    "South Africa": "ZAF"
}

rows = []

for country, code in countries_wb.items():
    url = f"https://api.worldbank.org/v2/country/{code}/indicator/NY.GDP.PCAP.CD?format=json&per_page=100"
    data = requests.get(url).json()[1]

    for row in data:
        if row["date"].isdigit() and 2000 <= int(row["date"]) <= 2024:
            rows.append({
                "Entity": country,
                "Year": int(row["date"]),
                "GDP per capita": row["value"]
            })

gdp5 = pd.DataFrame(rows)

print(gdp5.shape)
print(gdp5["Entity"].value_counts())
```

```python
merged = pd.merge(
    vdem5,
    gdp5,
    on=["Entity", "Year"],
    how="inner"
)

print("Dataset shape:", merged.shape)
print("\nObservations per country:")
print(merged["Entity"].value_counts())

print("\nMissing values:")
print(merged[["v2x_libdem", "GDP per capita"]].isna().sum())
```

```python
correlation = merged["v2x_libdem"].corr(
    merged["GDP per capita"]
)

print("Correlation between Liberal Democracy and GDP per capita:")
print(round(correlation, 3))
```

The analysis examined the association between GDP per capita and liberal democracy across the United States, China, Germany, Brazil, and South Africa from 2000 to 2024.

Across the full dataset of 125 country-year observations, GDP per capita and the Liberal Democracy Index had a moderate positive correlation (r = 0.548). At the country-average level, the correlation was also positive (r = 0.603).

However, the within-country correlations show that the relationship was not consistent over time. The correlations ranged from -0.919 in China to +0.390 in South Africa. These results suggest that the positive cross-national association between economic development and liberal democracy should not be interpreted as evidence that economic development automatically produces democratization. Countries can experience substantial economic growth while following very different political trajectories.

- Correlation does not establish causation.
- The analysis does not control for other factors that may influence democracy, such as institutions, political crises, historical legacies, or international factors.
- Within-country correlations are based on annual observations and capture only linear associations.

# Week 2: Democracy and Economic Development

## Research Question

How is economic development associated with liberal democracy across five prominent countries?

## Countries and Period

- United States
- China
- Germany
- Brazil
- South Africa
- 2000–2024

## Data

This project combines two datasets:

1. **V-Dem Core Dataset, Version 16**
   - Variable: `v2x_libdem`
   - Measures the Liberal Democracy Index.

2. **World Bank GDP per capita data**
   - Indicator: GDP per capita (current US$)
   - Retrieved through the World Bank API.

The two datasets were matched by country and year, producing 125 country-year observations.

## Method

The analysis uses:

- Data cleaning and country/year filtering
- Dataset merging
- Pearson correlation
- Country-level averages
- Within-country correlations
- Time-series and scatter plots

The analysis distinguishes between cross-national associations and changes occurring within individual countries over time.

## Key Findings

Across all 125 country-year observations, GDP per capita and liberal democracy had a moderate positive correlation:

**r = 0.548**

At the country-average level, the correlation was:

**r = 0.603**

However, the within-country results were much more heterogeneous:

| Country | Within-country correlation |
|---|---:|
| Brazil | 0.081 |
| China | -0.919 |
| Germany | -0.643 |
| South Africa | 0.390 |
| United States | -0.411 |

This shows that the positive cross-national association does not consistently appear within individual countries over time.

China is particularly notable: GDP per capita increased substantially during the period while its Liberal Democracy Index remained very low. Brazil showed almost no linear relationship, while South Africa had a moderate positive association.

## Interpretation

The findings suggest that richer countries in this five-country sample tended, on average, to have higher levels of liberal democracy. However, economic development does not appear to have a uniform relationship with democratization within countries.

The results therefore should not be interpreted as evidence that economic development causes democracy. Political institutions, historical trajectories, political crises, and other domestic and international factors may also shape democratic development.

## Limitations

- The analysis covers only five countries.
- The country-average correlation is based on only five observations.
- Correlation does not establish causation.
- The analysis does not control for other factors that may influence democracy.
- Within-country correlations capture linear relationships and should not be interpreted as causal effects.
- GDP per capita is measured in current US dollars and is not adjusted for inflation.

## Main Conclusion

Economic development and liberal democracy are positively associated across the countries examined, but the relationship is neither uniform nor deterministic. The contrast between the cross-national and within-country results demonstrates why economic development alone cannot explain differences in democratic trajectories.
